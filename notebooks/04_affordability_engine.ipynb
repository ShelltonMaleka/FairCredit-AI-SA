```python
# Add the project's source directory to Python's module search path
# so that the notebook can import the FairCredit affordability engine.

import sys
from pathlib import Path

project_root = Path.cwd().parent

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
```

```python
# Import the affordability assessment functions developed specifically
# for the FairCredit application.

from src.affordability_engine import (
    assess_affordability,
    minimum_expense_norm
)
```

```python
# Verify the minimum-expense calculation across several income bands.

test_incomes = [
    800,
    5_000,
    10_000,
    30_000,
    60_000
]

for income in test_incomes:
    expense = minimum_expense_norm(income)

    print(
        f"Gross income: R{income:,.2f} "
        f"| Minimum expense norm: R{expense:,.2f}"
    )
```

```python
# Example consumer profile used to confirm that the affordability
# engine produces a complete and understandable assessment.

result = assess_affordability(
    gross_income=10_000,
    statutory_deductions=1_200,
    declared_living_expenses=3_000,
    existing_debt_repayments=1_500,
    maintenance_obligations=300,
    proposed_installment=1_100
)

result
```

```python
# Present the affordability calculation in the same style that will
# eventually be displayed in the FairCredit web application.

print("FAIRCREDIT AFFORDABILITY ASSESSMENT")
print("-" * 45)

print(
    f"Gross income:                 "
    f"R{result.gross_income:,.2f}"
)

print(
    f"Net income:                   "
    f"R{result.net_income:,.2f}"
)

print(
    f"Minimum expense norm:         "
    f"R{result.minimum_expense_norm:,.2f}"
)

print(
    f"Living expenses used:         "
    f"R{result.living_expenses_used:,.2f}"
)

print(
    f"Existing debt repayments:     "
    f"R{result.existing_debt_repayments:,.2f}"
)

print(
    f"Other obligations:            "
    f"R{result.maintenance_obligations:,.2f}"
)

print("-" * 45)

print(
    f"Discretionary income:         "
    f"R{result.discretionary_income:,.2f}"
)

print(
    f"Proposed new instalment:      "
    f"R{result.proposed_installment:,.2f}"
)

print(
    f"Remaining after new credit:   "
    f"R{result.remaining_after_new_credit:,.2f}"
)

print("-" * 45)

print(
    "AFFORDABILITY:",
    "PASS" if result.affordability_pass else "FAIL"
)

print(
    "STATUS:",
    result.status
)
```
